```python
import scipy.io
import numpy as np
import os

data_dir = "/Users/isaaclee/Wildfire_Research/validation_case_data_and_preparation/test_fire_data"
output_folder = "/Users/isaaclee/Wildfire_Research/validation_case_data_and_preparation/test_fire_data_npy"

for filename in os.listdir(data_dir):
    if filename.endswith(".mat"):
        data = scipy.io.loadmat(f"{data_dir}/{filename}")
        data = data['data_vec']
        np.save(f'{output_folder}/{os.path.splitext(filename)[0]}.npy', data)
```
